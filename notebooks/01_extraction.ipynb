```python
import pandas as pd

df_dict = []

for column, meta in data_dictionary.items():
    df_dict.append({
        "column_name": column,
        "description": meta.get("description"),
        "type": meta.get("type"),
        "values": meta.get("values", "NA")
    })

data_dict_df = pd.DataFrame(df_dict)

data_dict_df
```
